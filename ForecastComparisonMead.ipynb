```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the Excel sheet
df = pd.read_excel('USBR24moLakeMead.xlsx', sheet_name='InflowGraph')

# Drop empty columns
df = df.dropna(axis=1, how='all')

# Set 'Trace' as index
df.set_index('Trace', inplace=True)

# Define inflow year columns
years = ['Year_1', 'Year_2', 'Year_3']
year_labels = ['2024 – 2025', '2025 – 2026', '2026 – 2027']

# Define custom styles
custom_styles = {
    'USBR Probable Minimums': {'color': 'red', 'linestyle': '-', 'linewidth': 3},
    'Glen Canyon Dam Release': {'color': 'red', 'linestyle': ':', 'linewidth': 3},
    'Powell inflow': {'color': 'pink', 'linestyle': '-', 'linewidth': 3},
    'Trace 1': {'color': 'lightblue', 'linestyle': '-', 'linewidth': 3},
    'Trace 2': {'color': 'darkblue', 'linestyle': '-', 'linewidth': 3}
}




# Plot each trace individually
plt.figure(figsize=(10, 6))
for trace, style in custom_styles.items():
    if trace in df.index:
        # Insert line breaks manually for long labels
        wrapped_label = trace.replace('Glen Canyon Dam Release', 'Glen Canyon\nDam Release') \
                             .replace('USBR Probable Minimums', 'USBR Probable \nMinimums')
        plt.plot(year_labels, df.loc[trace, years], label=wrapped_label, **style)


# Customize plot
#plt.title('Mead Inflow: Selected Traces and Scenarios')
plt.ylabel('Flow MAF/year \n(Million Acre-Feet Per Year)')
plt.rcParams.update({'font.size': 24})
plt.xticks(fontsize=20)               
plt.yticks(fontsize=20)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=16)
#plt.xlabel('Year')
#plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()



```
