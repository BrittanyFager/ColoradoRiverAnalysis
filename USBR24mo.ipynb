```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the Excel file
df = pd.read_excel("C:\Users\Fager\Documents\LakeMead\USBR24moLakeMead.xlsx, sheet_name="Mead24mMin")

# Convert 'Date' column to datetime format
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

# Drop rows with missing dates (like WY summary rows)
df = df.dropna(subset=['Date'])

# Plot Reservoir Elevation over Time
plt.figure(figsize=(12, 6))
plt.plot(df['Date'], df['ReservoirElevationEndofMonth(Ft)'], marker='o', linestyle='-')
plt.title("Lake Mead End-of-Month Reservoir Elevation")
plt.xlabel("Date")
plt.ylabel("Elevation (Feet)")
plt.grid(True)
plt.tight_layout()
plt.show()
```
